```python
import tkinter as tk

# Light state management (0: Red, 1: Green, 2: Yellow)
state = 0  

def change_light():
    global state
    
    # Reset all lights to their dim/off states instantly
    lbl_red.config(bg="#330000")
    lbl_yellow.config(bg="#333300")
    lbl_green.config(bg="#003300")
    
    if state == 0:
        lbl_red.config(bg="#FF0000")       # Turn on Red
        lbl_status.config(text="STOP", fg="#FF0000")
        state = 1
        
    elif state == 1:
        lbl_green.config(bg="#00FF00")     # Turn on Green
        lbl_status.config(text="GO", fg="#00FF00")
        state = 2
        
    elif state == 2:
        lbl_yellow.config(bg="#FFFF00")    # Turn on Yellow
        lbl_status.config(text="SLOW DOWN", fg="#FFFF00")
        state = 0

# Main window setup optimized for lightning-fast loading
root = tk.Tk()
root.title("Fast Traffic Light")
root.geometry("180x360")
root.config(bg="#111111")

# Lightweight frame box for the lights
frame = tk.Frame(root, bg="#222222", bd=5, relief="ridge")
frame.pack(pady=15)

# Fast-rendering square labels (No heavy canvas engine used)
lbl_red = tk.Label(frame, width=6, height=2, bg="#FF0000")
lbl_red.pack(pady=5, padx=15)

lbl_yellow = tk.Label(frame, width=6, height=2, bg="#333300")
lbl_yellow.pack(pady=5, padx=15)

lbl_green = tk.Label(frame, width=6, height=2, bg="#003300")
lbl_green.pack(pady=5, padx=15)

# Status indicator and action button
lbl_status = tk.Label(root, text="STOP", font=("Arial", 12, "bold"), fg="#FF0000", bg="#111111")
lbl_status.pack()

btn = tk.Button(root, text="CHANGE", font=("Arial", 10, "bold"), bg="#444444", fg="white", bd=0, padx=10, pady=5, command=change_light)
btn.pack(pady=15)

root.mainloop()
```

```python
import turtle
import cv2
import numpy as np
import sys
import os

# 1. Image Path Check
image_path = r"D:\Python Practice Tbrain\ganesh.jpg"

# Safety Check: Verify if the file actually exists at the path
if not os.path.exists(image_path):
    print(f"❌ Error: File not found at this path -> {image_path}")
    print("Please check the file name, extension (.jpg/.png), and folder path.")
    sys.exit()

img = cv2.imread(image_path)
if img is None:
    print(f"❌ Error: OpenCV could not decode the image. The file might be corrupted.")
    sys.exit()

# 2. Resize image
target_width = 650
h, w = img.shape[:2]
target_height = int((h / w) * target_width)
img_resized = cv2.resize(img, (target_width, target_height), interpolation=cv2.INTER_AREA)

# 3. Convert to grayscale
gray = cv2.cvtColor(img_resized, cv2.COLOR_BGR2GRAY)

# 4. Create binary image
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY_INV)

# 5. Find contours
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)

# 6. Setup turtle canvas
try:
    turtle.clearscreen() # Clears any frozen previous windows
except:
    pass

screen = turtle.Screen()
screen.setup(width=850, height=850)
screen.bgcolor("#181B22")
screen.title("Ganesha Metallic Gold Clean Animation")
screen.tracer(0, 0) # Set to 0 to prevent lag and render instantly

# 7. Create turtle pen
pen = turtle.Turtle()
pen.hideturtle()
pen.speed(0)
pen.pensize(1.5)

# 8. Colors
gold_stroke = "#D49B24"

# 9. Image offset (To center the drawing)
offset_x = target_width / 2
offset_y = target_height / 2

# 10. Convert OpenCV point array to turtle coordinates safely
def convert_point(point):
    # OpenCV stores points as nested arrays: [[x, y]]
    pt = point[0]
    x = pt[0]
    y = pt[1]
    
    turtle_x = x - offset_x
    turtle_y = offset_y - y # Flips Y-axis so it doesn't draw upside down
    return turtle_x, turtle_y

# 11. Draw one contour
def draw_contour(contour):
    if len(contour) < 2:
        return
    
    first_x, first_y = convert_point(contour[0])
    pen.penup()
    pen.goto(first_x, first_y)
    pen.pendown()
    
    for point in contour[1:]:
        x, y = convert_point(point)
        pen.goto(x, y)
    
    pen.penup()

# 12. Sort contours by size (Draws main details first)
contours = sorted(contours, key=cv2.contourArea, reverse=True)

# 13. Draw ganesha
print("Starting Ganesha animation...")
pen.color(gold_stroke)

for contour in contours:
    area = cv2.contourArea(contour)
    if area < 10: # Filters out tiny background noise/dots
        continue
    draw_contour(contour)

# 14. Finish & Render
screen.update() # Forces the screen to update and display the complete image
print("Ganesha drawing completed!")

# Safest loop close for VS Code / Jupyter Notebook environments
try:
    turtle.mainloop()
except KeyboardInterrupt:
    print("Program closed by user.")
```

```python
numbers = (10, 25, 10, 40, 55, 25, 70, 40, 80)

# Total elements
print("Total Elements:", len(numbers))

# Unique values
unique = ()

for n in numbers:
    if n not in unique:
        unique += (n,)

print("Unique Values:", unique)

# Even and Odd
even = 0
odd = 0

for n in numbers:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)

# Largest
largest = numbers[0]

for n in numbers:
    if n > largest:
        largest = n

print("Largest:", largest)

# Smallest
smallest = numbers[0]

for n in numbers:
    if n < smallest:
        smallest = n

print("Smallest:", smallest)
```

```python
nums = [2, 3, 4, 5, 8, 7,9,4,6,20,4]
target = 7
def find_pairs(nums, target):
    pairs = []
    
    for i in range(len(nums)):
        for j in range(i +1,len(nums)):
            if nums[i] + nums[j] == target:
                pairs.append((nums[i], nums[j]))
    return pairs
print(find_pairs(nums, target))        
```

```python
class Atm:

  # constructor(special function)->superpower -> 
  def __init__(self):
    print(id(self))
    self.pin = ''
    self.balance = 0
    #self.menu()

  def menu(self):
    user_input = input("""
    Hi how can I help you?
    1. Press 1 to create pin
    2. Press 2 to change pin
    3. Press 3 to check balance
    4. Press 4 to withdraw
    5. Anything else to exit
    """)

    if user_input == '1':
      self.create_pin()
    elif user_input == '2':
      self.change_pin()
    elif user_input == '3':
      self.check_balance()
    elif user_input == '4':
      self.withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin = input('enter your pin')
    self.pin = user_pin

    user_balance = int(input('enter balance'))
    self.balance = user_balance

    print('pin created successfully')
    self.menu()

  def change_pin():
    old_pin = input('enter old pin')

    if old_pin == self.pin:
      # let him change the pin
      new_pin = input('enter new pin')
      self.pin = new_pin
      print('pin change successful')
      self.menu()
    else:
      print('nai karne de sakta re baba')
      self.menu()

  def check_balance(self):
    user_pin = input('enter your pin')
    if user_pin == self.pin:
      print('your balance is ',self.balance)
    else:
      print('chal nikal yahan se')

  def withdraw(self):
    user_pin = input('enter the pin')
    if user_pin == self.pin:
      # allow to withdraw
      amount = int(input('enter the amount'))
      if amount <= self.balance:
        self.balance = self.balance - amount
        print('withdrawl successful.balance is',self.balance)
      else:
        print('abe garib')
    else:
      print('sale chor')
    self.menu()


```
